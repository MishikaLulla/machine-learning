# Support Vector Machine
Support Vector Machine is a supervised machine learning algorithm mainly used for classification
The main idea is to find the best boundary or hyperplane that separates different classes

# Support Vectors
The data points closest to the decision boundary are called Support Vectors
They are very important because they help determine where the boundary should be placed

# Margin
The distance between the decision boundary and the nearest support vectors is called the margin
SVM tries to find the boundary with the maximum possible margin

# Optimal Hyperplane

# Hyperplane

In simple cases with 2 features it is a line
With more dimensions it can be a plane or a higher dimensional boundary

# How SVM Works  Margin and Support Vectors
* 1  Margin
Margin is the distance between the hyperplane and the closest data points from each class
SVM tries to maximize this margin

Why
A larger margin generally helps the model perform better on unseen data and makes the model more robust

Larger margin → better generalization

* 2  Support Vectors
Support vectors are the data points that are closest to the hyperplane
They are very important because they support and determine the position of the decision boundary
ex- the hyperplane as a road and support vectors as the points at the edges of the road
If a support vector moves the hyperplane can also move

               \             /
                \  Margin   /
                 \         /
                  ---------
                  Hyperplane


# advantages of Support Vector Machine SVM

 1  Effective in High Dimensions
SVM performs well when the dataset has many features or dimensions
So it can be useful for complex datasets with a large number of features

2  Memory Efficient
SVM mainly uses the support vectors to define the decision boundary
Because of this it can be relatively memory efficient during prediction

3  Versatile
SVM can handle different types of data and relationships
The kernel trick allows SVM to handle relationships ranging from linear to highly non linear patterns

4  Robust
SVM is generally less prone to overfitting in high dimensional spaces, especially when appropriate regularization is used.


# Limitations of Support Vector Machine SVM

* 1  Slow Training
SVM can be slow to train on very large datasets because of its computational complexity.
As the dataset becomes bigger, training can take more time.

* 2  Hyperparameter Tuning
SVM performance depends heavily on choosing the right kernel and hyperparameters.
Important parameters include
C
Gamma
and the choice of kernel
So careful tuning may be required to get good performance.

* 3  Does Not Directly Provide Probability Estimates
SVM does not directly provide probability estimates.
Because of this, it is less suitable when we specifically need probabilistic outputs.

* 4  Does Not Scale Well to Very Large Datasets
SVM may not work efficiently with datasets containing millions of examples.
As the dataset size increases, training time can increase significantly


Slow training → Difficult with very large datasets
Tuning → Kernel and hyperparameters need careful selection
No direct probability → Doesn't directly give probability estimates
Large datasets → Doesn't scale well to millions of examples
