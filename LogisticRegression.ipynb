* Evaluation Metrics for Classification


1 Accuracy
Accuracy tells us the overall correctness of the model.

Formula

Accuracy = (TP + TN) / Total
Where
TP = True Positive
TN = True Negative
Example
If the model makes 100 predictions and 90 are correct
Accuracy = 90%

Important: Accuracy can be misleading when the dataset is imbalanced.

2 Precision
Precision tells us
When the model predicts positive, how often is it actually correct?

Formula

Precision = TP / (TP + FP)
Example

Model predicts 10 patients have a disease

Actually only 8 have the disease

Precision = 8 / 10 = 80%
So precision focuses on false positives.

3 Recall

Recall tells us
Out of all the actual positive cases, how many did the model correctly find?

Formula
Recall = TP / (TP + FN)
Example

There are actually 10 patients with the disease

Model correctly detects 8

Recall = 8 / 10 = 80%
So recall focuses on false negatives.

4 ROC Curve

ROC stands for Receiver Operating Characteristic.
It is used to evaluate a binary classification model at different classification thresholds.

It compares
True Positive Rate → Recall

against
False Positive Rate
The ROC curve helps us understand how well the model separates the positive and negative classes.
