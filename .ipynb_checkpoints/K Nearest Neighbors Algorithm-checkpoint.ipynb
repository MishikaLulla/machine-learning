# K Nearest Neighbors 
 a supervised Machine Learning algorithm used for both classification and regression but it is commonly used for classification
 # How KNN works
Suppose we want to classify a fruit as an apple or an orange based on its features
Choose the value of K
Calculate the distance between the new data point and existing data points
Find the K nearest neighbors
For classification-select the majority class among those neighbors
Assign that class to the new data point

# K
K represents the number of nearest neighbors considered when making a prediction
K = 1: considers only the closest point
K = 3 : considers the three closest points
K = 5 considers the five closest points

# Feature scaling
Feature scaling is important in KNN because it depends on distances
For example, if one feature ranges from 1 to 10 and another ranges from 1,000 to 100,000, the larger-scale feature may dominate the distance calculation
We can use StandardScaler to bring the features to comparable scales.
