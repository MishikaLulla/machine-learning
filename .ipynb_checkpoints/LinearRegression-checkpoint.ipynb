# Linear Regression 
* Regression
Regression is a Supervised Machine Learning technique used to predict a continuous numerical value.
In simple words
it predicts how much or how many
Examples
Predicting house price
Predicting student's marks
Predicting temperature
Predicting sales

* Linear Regression

Linear Regression is a regression algorithm that finds a straight line relationship between input and output.
For example
If the number of hours studied increases, the marks may also increase
The model finds the best fitting line through the data and uses it to predict marks for a new number of study hours

y = b₀ + b₁x

where

y = predicted value
x = input value
b₀ = intercept
b₁ = slope


Use case of Linear Regression

House Price Prediction

Input can be

House size

Number of rooms

Location related numerical features

Output

Predicted house price

So if we provide the size of a new house, Linear Regression can estimate its price based on the relationship learned from previous house data
