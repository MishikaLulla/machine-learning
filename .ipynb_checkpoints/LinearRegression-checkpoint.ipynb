# Linear Regression 
* Regression
Regression is a Supervised Machine Learning technique used to predict a continuous numerical value.
In simple words
it predicts how much or how many
Examples
Predicting house price
Predicting student's marks
Predicting temperature
Predicting sales

* Linear Regression

Linear Regression is a regression algorithm that finds a straight line relationship between input and output.
For example
If the number of hours studied increases, the marks may also increase
The model finds the best fitting line through the data and uses it to predict marks for a new number of study hours

y = b₀ + b₁x
(y=mx+c)
where

y = predicted value
x = input value
b₀ = intercept
b₁ = slope


Use case of Linear Regression

House Price Prediction

Input can be

House size

Number of rooms

Location related numerical features

Output

Predicted house price

So if we provide the size of a new house, Linear Regression can estimate its price based on the relationship learned from previous house data


* Ordinary Least Squares OLS

Ordinary Least Squares is the method used to find the best fitting line in Linear Regression.

In simple words
OLS tries to find a line that is as close as possible to all the actual data points
For every data point there is a difference between

Actual value
and
Predicted value
This difference is called the residual or error
OLS finds the line for which the sum of squared errors is minimum



Why do we square the errors?

Errors can be positive or negative

For example
Actual = 10
Predicted = 8
Error = 2

Actual = 10
Predicted = 12
Error = -2

If we add them they can cancel each other
So we square them

2² = 4
(-2)² = 4

Then we add all the squared errors
OLS chooses the line with the minimum total squared error

* Assumptions of Linear Regression

These are the conditions we assume before applying Linear Regression

1 Linearity
There should be a linear relationship between the input variable and output variable

Example
As study hours increase, marks should increase in a roughly straight line pattern

2 Independence
The observations should be independent of each other
The value of one observation should not affect another observation

3 No Multicollinearity

The independent variables should not be highly related to each other
For example, if we use house size in square feet and house size in square meters together, they provide almost the same information

4 Homoscedasticity
The errors should have constant variance across different predicted values
In simple words, the spread of errors should remain roughly the same

5 Normality of Errors
The errors or residuals should be approximately normally distributed
This assumption is mainly important for statistical tests and confidence intervals


* Model Evaluation Metrics for Regression

Accuracy-Accuracy tells us how many predictions the model got correct out of all predictions

Formula
Accuracy = Correct Predictions / Total Predictions

Example
If the model makes 100 predictions and 90 are correct
Accuracy = 90 percent

Accuracy tells us the overall percentage of correct predictions.

2 Precision
Precision tells us
Out of all the cases predicted as positive how many were actually positive

Example
A model predicts 10 emails as spam
Actually only 8 are spam
Precision = 8 / 10 = 80 percent
High precision means fewer false positives

Easy definition
Precision focuses on how correct the positive predictions are.

3 Recall
Recall tells us
Out of all the actual positive cases how many did the model correctly identify
Example
There are actually 10 spam emails
The model correctly identifies 8
Recall = 8 / 10 = 80 percent

High recall means fewer false negatives
Easy definition

Recall focuses on finding as many actual positive cases as possible.

4 F1 Score
F1 Score combines Precision and Recall into one metric.
It is useful when we want a balance between precision and recall.

Formula
F1 Score = 2 × Precision × Recall / Precision + Recall

A higher F1 Score generally means a better balance between precision and recall.

After training a Linear Regression model we need to check how well the model is performing For this we use evaluation metrics


1-----R² Score

R² is called the coefficient of determination.
It tells us how much of the variation in the output is explained by the model.
Its value is usually between 0 and 1

Example
R² = 0.80
This means the model explains about 80 percent of the variation in the dependent variable.
Higher R² generally means better fit.
Easy definition
R² tells us how well the independent variables explain the dependent variable


2-----Adjusted R²

Adjusted R² is an improved version of R²
The problem with R² is that it can increase when we add more independent variables even if those variables are not useful
Adjusted R² takes the number of variables into account and only improves when a new variable provides useful information

Example
Model 1 → R² = 0.85
Model 2 → R² = 0.86

If the extra variable is not actually useful, Adjusted R² may stay the same or even decrease.

Adjusted R² measures model performance while considering the number of independent variables used in the model


3------Error Metrics

Error metrics tell us how far the predicted values are from the actual values.
Common error metrics are
MAE — Mean Absolute Error

It calculates the average absolute difference between actual and predicted values.

Example
Actual marks = 80
Predicted marks = 75
Error = 5

Lower MAE means better performance
